```python
# I want to see how many of each species were sighted at each sighting (each row). 

#how many sighted of each species
species_count = pfw.groupby(['american_english_name', 'how_many']).agg(
    highest_sighting = ('how_many', 'max')

).sort_values(by='how_many', ascending=False)

## This was an alternate way to count them: pfw[['american_english_name', 'how_many']].sort_values(by=['how_many'],ascending=False)

## I tried it first with just sort_values() and then with groupby().sort_values() to check that 500 was correct. 
## Using .groupby(), I can see that there are only two species that have a sighting of 500 birds, which seems like too round of a number and very high. 
## I need to look for outliers and likely remove them from the data, since I am trying to find the average best feeder placement. 
  
```

```python
# Since there are already clearly outliers, I want to make sure that all of the sighting data is accurate. 
# There is a 'reviewed' column that should indicate the review state of each observation based on the PFW flagging system.
# Of the 504985 rows, only 2038 have been reviewed. 
pfw['reviewed'].value_counts()
```

```python

# I also wanted to check the 'valid' column, which indicates whether the observation is considered valid by the PFW flagging system. 
# Sightings that are not valid cannot be verified as true and/or accurate, and so are not useful data. 
# Of the 504985 rows, 3674 are valid. 
pfw['valid'].value_counts()
```

```python
# I want to create a dataframe that has all of the rows and columns ONLY if 'valid' and 'reviewed' are True (1.0)

pfw_VR = pfw.loc[(pfw['valid']==1.0) & (pfw['reviewed']==1.0)]
pfw_VR

#Now the pfw_VR (validated and reviewed) dataframe has only entries that are confirmed by PFW to be true sightings. I will work with this data from here on. 
```

```python
#I will also pare down the dataframe some more based on checkpoint 3, using .drop() and .query() to remove unnecessary rows from the 'pfw_species_codes_dict.csv' and data that are not from the year 2021. 
pfw_VR = pfw_VR.query('proj_period_id == "PFW_2021"')
pfw_VR = pfw_VR.drop(columns=['alt_full_spp_code','n_locations', 'taxonomy_version', 'taxonomic_sort_order'])
pfw_VR = pfw_VR.drop(columns=['subnational1_code', 'entry_technique', 'sub_id', 'obs_id', 'Data_Entry_Method'])
pfw_VR


```

```python
pfw_VR.shape

# I now have 282 rows and 80 columns to work with, and my data is limited to validated and reviewed sightings from 2021. 
# I will now explore more of the columns that describe the feeder locations, weather conditions, and predators. 
```

```python
bird_agg_dict = {
    'how_many': 'sum',                # Summing the number of birds
    'species_code': 'nunique' # Count unique bird types 
}
most_common_bird_types = pfw_VR.groupby('american_english_name').agg(
    bird_agg_dict
).sort_values(by='how_many', ascending=False)

most_common_bird_types
```

```python
# I want to see what type of birds used each type of feeder. I will start with the suet feeder, summing the unique birds and how many came to suet feeders, as well as how many feeders. 

suet_agg_dict = {
    'how_many': 'sum',                # Summing the number of birds using the suet feeder
    'numfeeders_suet': 'count',         # Counting the number of suet
    'species_code': 'nunique' # Count unique bird types 
}

suet_feeder_bird_types = pfw_VR.groupby('american_english_name').agg(
    suet_agg_dict
).sort_values(by='how_many', ascending=False)

with pd.option_context('display.max_rows', None, 
                       'display.max_columns', None, 
                       'display.precision', 3):
    print(suet_feeder_bird_types)
```

```python
#When trying the same code with the ground feeders, I got exactly the same result, which means I am not counting the correct thing. 
ground_agg_dict = {
    'how_many': 'sum',                # Summing the number of birds using the ground feeder
    'numfeeders_ground': 'count',         # Counting the number of ground feeders
    'species_code': 'nunique' # Count unique bird types 
}

ground_feeder_bird_types = pfw_VR.groupby('american_english_name').agg(
    ground_agg_dict
).sort_values(by='how_many', ascending=False)

with pd.option_context('display.max_rows', None, 
                       'display.max_columns', None, 
                       'display.precision', 3):
    print(ground_feeder_bird_types)
```

## Visualize

```python
import matplotlib.pyplot as plt
```

Getting a sense for outliers

```python
#I will use a box and whisker plot to get a sense of outlier sightings that have a very large number of birds. 

x = [pfw_VR['how_many']]

plt.boxplot(x)
```

Exploring the feeder conditions 

```python
# I wanted to determine which feeders were the most popular to use, and therefore would be the most useful for my data.
# Since hanging, water, and thistle feeders have very few uses, I am going to remove them from the rest of my analysis. 
print(pfw_VR['numfeeders_suet'].isna().value_counts(),
      pfw_VR['numfeeders_ground'].isna().value_counts(),
      pfw_VR['numfeeders_hanging'].isna().value_counts(),
      pfw_VR['numfeeders_platfrm'].isna().value_counts(),
      pfw_VR['numfeeders_humming'].isna().value_counts(),
      pfw_VR['numfeeders_water'].isna().value_counts(),
      pfw_VR['numfeeders_thistle'].isna().value_counts(),
      pfw_VR['numfeeders_fruit'].isna().value_counts(),
      pfw_VR['numfeeders_hopper'].isna().value_counts(),
      pfw_VR['numfeeders_tube'].isna().value_counts(),
      pfw_VR['numfeeders_other'].isna().value_counts()
      )

columns = ['numfeeders_suet', 
           'numfeeders_ground', 
           'numfeeders_hanging', 
        'numfeeders_platfrm', 
        'numfeeders_humming', 
        'numfeeders_water', 
        'numfeeders_thistle', 
        'numfeeders_fruit', 
        'numfeeders_hopper', 
        'numfeeders_tube', 
        'numfeeders_other']

num_of_feeders = {}

for col in columns:
    num_of_feeders[col] = (~pfw_VR[col].isna()).sum()

num_of_feeders_series = pd.Series(num_of_feeders)

plt.figure(figsize=(10,6))
num_of_feeders_series.plot(kind='bar', color='orange')
plt.title('Count of Types of Feeders')
plt.xlabel('Feeder Type')
plt.ylabel('Number of Entries')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
#I will do a similar check on the types of habitats and other variables to get the most complete data for my analysis. 
columns = [
    'hab_dcid_woods', 
    'hab_evgr_woods',
    'hab_mixed_woods',
    'hab_orchard', 
    'hab_park', 
    'hab_water_fresh', 
    'hab_water_salt', 
    'hab_residential', 
    'hab_industrial', 
    'hab_agricultural',
    'hab_desert_scrub', 
    'hab_young_woods', 
    'hab_swamp', 
    'hab_marsh'
]

type_of_habitat = {}

for col in columns:
    type_of_habitat[col] = (~pfw_VR[col].isna()).sum()

type_of_habitat_series = pd.Series(type_of_habitat)

plt.figure(figsize=(10,6))
type_of_habitat_series.plot(kind='bar', color='skyblue')
plt.title('Count of Types of Habitats')
plt.xlabel('Habitat Type')
plt.ylabel('Number of Entries')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# There is a fairly even spread of entries for each type of habitat. I will likely need to choose a different variable if I want to pare down the data. 
# Or, the many types of habitat could provide interesting comparisons. 
```

```python
# Visualizing the yard types

columns = [
    'yard_type_pavement', 
    'yard_type_garden', 
    'yard_type_landsca', 
    'yard_type_woods', 
    'yard_type_desert'
]

type_of_yard = {}

for col in columns:
    type_of_yard[col] = (~pfw_VR[col].isna()).sum()

type_of_yard_series = pd.Series(type_of_yard)

plt.figure(figsize=(10,6))
type_of_yard_series.plot(kind='bar', color='lightgreen')
plt.title('Count of Types of Yards')
plt.xlabel('Yard Type')
plt.ylabel('Number of Entries')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# There is again a fairly even spread of different yard types.  
```
