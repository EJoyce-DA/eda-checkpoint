```python
# I want to see what type of birds were sighted at each type of feeder. 
# I will start with the suet feeder and try to see how many birds of each species were observed at suet feeders 

suet_species = (
    pfw_VR
    .loc[pfw_VR['numfeeders_suet'] > 0] #filter so only rows with at least one suet feeder present are counted
    .groupby('american_english_name')
    .agg(
        observations=('american_english_name', 'size'), #number of rows per species
        total_birds=('how_many', 'sum') #total number of birds observed per species
    )
    .sort_values(by='observations', ascending=False)
)

suet_species.head(10)
```

```python
# I want to continue this line of analysis for the other feeder types, so I will create a dictionary and a loop to repeat this process for all of the numfeeder columns 

feeder_cols = [
    'numfeeders_suet', 'numfeeders_ground', 'numfeeders_hanging', 'numfeeders_platfrm',
    'numfeeders_humming', 'numfeeders_water', 'numfeeders_thistle',
    'numfeeders_fruit', 'numfeeders_hopper', 'numfeeders_tube',
    'numfeeders_other'
]

feeder_species_stats = {}

for feeder in feeder_cols:
    feeder_species_stats[feeder] = (
        pfw_VR
        .loc[pfw_VR[feeder] > 0]  # only rows where this feeder type is present
        .groupby('american_english_name')
        .agg(
            observations=('american_english_name', 'size'),
            total_birds=('how_many', 'sum')
        )
        .sort_values(by='observations', ascending=False)
    )


feeder_species_stats['numfeeders_humming'].head(10)   #I can change the column that is displayed to view the results for other columns

# This is showing me 
```

```python
# I now want to see how many unique species (species richenss) were sighted at each feeder type
# I will use a loop to find the unique species sightings for each feeder type, and then use sort_values() to order them 

feeder_cols = [
    'numfeeders_suet',
    'numfeeders_ground',
    'numfeeders_hanging',
    'numfeeders_platfrm',
    'numfeeders_humming',
    'numfeeders_water',
    'numfeeders_thistle',
    'numfeeders_fruit',
    'numfeeders_hopper',
    'numfeeders_tube',
    'numfeeders_other'
]

unique_species_by_feeder = {}

for feeder in feeder_cols:
    unique_species_by_feeder[feeder] = (
        pfw_VR.loc[pfw_VR[feeder] > 0, 'american_english_name']
        .nunique()
    )

unique_species_by_feeder = (
    pd.Series(unique_species_by_feeder)
    .sort_values(ascending=False)
)

unique_species_by_feeder.reset_index()

# This indicates that suet feeders have the highest species richness, while water feeders have the lowest species richness. 
```

```python
# I have looked at the type of feeders, but now I want to explore if the number of feeders has an impact on the number of birds. 
# I will create a site summary that tells me the sum of birds and the sum of feeders for each location

feeder_cols = [c for c in pfw_VR.columns if c.startswith('numfeeders_')]

site_summary = (
    pfw_VR
    .groupby('loc_id') #grouped by each observation site
    .agg(
        total_birds=('how_many', 'sum') #number of individual bords observed at each site
    )
)

# Sum feeders per location (row-wise)
site_summary['total_feeders'] = (
    pfw_VR
    .groupby('loc_id')[feeder_cols]
    .sum() #sum of each feeder type per location
    .sum(axis=1) # sum across feeder types (total number of feeders at each site)
)
site_summary.head(15)

# This summary could be used to find a trend related to the number of feeders vs the number of birds. I will visualize this further down. 
```

```python
# I want to know the average number of birds that visit each feeder type. 
# I will use a list of the feeder columns and create a dictionary to contain the average birds per feeder type. 
# I will then use a loop to calculate the average for each feeder type

# List of feeder columns
feeder_cols = [
    'numfeeders_suet',
    'numfeeders_ground',
    'numfeeders_hanging',
    'numfeeders_platfrm',
    'numfeeders_humming',
    'numfeeders_water',
    'numfeeders_thistle',
    'numfeeders_fruit',
    'numfeeders_hopper',
    'numfeeders_tube',
    'numfeeders_other'
]

#Sum of birds per sighting
site_bird_totals = (
    pfw_VR.groupby('loc_id')['how_many']
    .sum()
    .rename('total_birds')
)

#sum of feeders per sighting
site_feeders = (
    pfw_VR.groupby('loc_id')[feeder_cols].sum()
)

site_feeder_totals_df = site_bird_totals.to_frame().join(site_feeders).reset_index()

# Dictionary to store average total birds per feeder type
avg_birds_per_feeder = {}

for feeder in feeder_cols:
    # Only consider sites with at least 1 of this feeder type
    avg_birds = site_feeder_totals_df[site_feeder_totals_df[feeder] > 0]['total_birds'].mean()
    avg_birds_per_feeder[feeder] = avg_birds


avg_birds_per_feeder = pd.Series(avg_birds_per_feeder).sort_values(ascending=False)

avg_birds_per_feeder.reset_index()

```

Impact of other factors

```python
# I am also interested in how certain other factors impact the 
# I will look at the following factors: water source, bird baths, nearby feeders, squirrels, cats, dogs, humans, and housing density

# First, I will create a groupby to compute the species richness for each loc_id
# species richness = unique species per location
site_species_richness = (
    pfw_VR
    .groupby('loc_id')['american_english_name']
    .nunique()
    .rename('species_richness')
)

# I then need a list for the different factors I am looking at and will create a groupby to look at the loc_ids that have the highest presence of these variables.
# Variables of interest
vars_of_interest = [
    'water_srcs_atleast',
    'bird_baths_atleast',
    'nearby_feeders',
    'squirrels',
    'cats',
    'dogs',
    'humans',
    'housing_density'
]

site_vars = (
    pfw_VR
    .groupby('loc_id')[vars_of_interest]
    .max()  
)

# I then created a new dataframe that combines the variables with species richness 
site_richness_df = site_species_richness.to_frame().join(site_vars).reset_index()

site_richness_df

# This dataframe shows the speceis richenss per loc_id with the other variables represented in each row. I can now use this dataframe for visualizations 
```

```python
# I am also interested in the total bird counts for each sighting with each variable, rather than the unique species. 
site_bird_totals = (
    pfw_VR
    .groupby('loc_id')['how_many']
    .sum()
    .rename('total_birds')
)


vars_of_interest = [
    'water_srcs_atleast',
    'bird_baths_atleast',
    'nearby_feeders',
    'squirrels',
    'cats',
    'dogs',
    'humans',
    'housing_density'
]

site_vars = (
    pfw_VR
    .groupby('loc_id')[vars_of_interest]
    .max()  
)

site_bird_totals_df = site_bird_totals.to_frame().join(site_vars).reset_index()

site_bird_totals_df
```

```python
# I want to know the average number of birds that are sighted at locations where other variables are present
# Variables of interest
vars_of_interest = [
    'water_srcs_atleast',
    'bird_baths_atleast',
    'nearby_feeders',
    'squirrels',
    'cats',
    'dogs',
    'humans',
    'housing_density'
]

avg_birds_present = {}

for var in vars_of_interest:
    # Sites where the variable is present (>0)
    avg_birds = site_bird_totals_df[site_bird_totals_df[var] > 0]['total_birds'].mean()
    avg_birds_present[var] = avg_birds


avg_birds_present = pd.Series(avg_birds_present).sort_values(ascending=False)
avg_birds_present

# Sites with cats had the highest average number of birds. 
# Since cats are predators, I would guess that they would lower the number of birds. 
# I'm interested if this indicates that cats seek out bird feeders and thus a higher number of sightings have cats present.
```

```python
# I wanted to try plotting the average number of birds as a bar plot to see what that looks like.
# It shows that when there are other variables present (1), the average number of birds can be different. 
# Squirrels had a lower number of average birds present than cats, for example, even though cats are predators. 
# I am interested if this indicates that squirrles cause an issue by trying to access the same feeders. 

vars_of_interest = [
    'water_srcs_atleast',
    'bird_baths_atleast',
    'nearby_feeders',
    'squirrels',
    'cats',
    'dogs',
    'humans',
    'housing_density'
]

avg_birds_present = {}

for var in vars_of_interest:
    # Sites where the variable is present (>0)
    avg_birds = site_bird_totals_df[site_bird_totals_df[var] > 0]['total_birds'].mean()
    avg_birds_present[var] = avg_birds

avg_birds_present = pd.Series(avg_birds_present).sort_values(ascending=False)

vars_labels_dict = {
     'water_srcs_atleast': 'Water Source',
    'bird_baths_atleast': 'Bird Bath',
    'nearby_feeders': 'Nearby Feeders',
    'squirrels': 'Squirrels',
    'cats': 'Cats',
    'dogs': 'Dogs',
    'humans': 'Humans',
    'housing_density': 'Housing Density'
}

vars_labels = avg_birds_present.index.map(vars_labels_dict)
vars_values = avg_birds_present.values


plt.figure(figsize=(10,6))
plt.bar(vars_labels, vars_values, color='thistle')
for i, val in enumerate(avg_birds_present.values):
    plt.text(i, val + 0.5, f'{val:.1f}', ha='center', va='bottom')
plt.ylabel('Average Total Birds')
plt.title('Average Total Birds at Sites Where Variable is Present')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
